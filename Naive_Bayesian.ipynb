```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# 🔹 Step 1: Create Sample Dataset (Spam & Ham Messages)
data = {
    'text': [
        'Congratulations! You have won a free trip to Paris!',
        'Limited time offer! Buy now and get 50% off.',
        'Hey, can we reschedule our meeting for tomorrow?',
        'Click the link to claim your lottery prize!',
        'Your bill payment is due tomorrow, please check your email.',
        'Win a brand new smartphone! Act fast before it expires.',
        'Let’s catch up for lunch this weekend!',
        'Claim your exclusive reward today! Offer expires soon.',
        'Your bank statement is available for download.',
        'Reminder: Your doctor’s appointment is tomorrow at 3 PM.'
    ],
    'label': ['spam', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham', 'ham']
}

# 🔹 Step 2: Convert to DataFrame
df = pd.DataFrame(data)

# 🔹 Step 3: Encode Labels (spam = 1, ham = 0)
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

# 🔹 Step 4: Split into Training and Testing Sets (70% Train, 30% Test)
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.3, random_state=42)

# 🔹 Step 5: Convert Text to Numerical Features using CountVectorizer
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)  # Fit on training data
X_test_vec = vectorizer.transform(X_test)  # Transform test data

# 🔹 Step 6: Train Naïve Bayes Classifier
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# 🔹 Step 7: Make Predictions
y_pred = model.predict(X_test_vec)

# 🔹 Step 8: Evaluate Model Performance
print(" Accuracy:", accuracy_score(y_test, y_pred))
print("\n Classification Report:\n", classification_report(y_test, y_pred, zero_division=1))

# 🔹 Step 9: Test with a Custom Message
test_message = ['You have been selected for a free gift card! Click here to claim.']
test_vec = vectorizer.transform(test_message)
prediction = model.predict(test_vec)[0]
print("\n Prediction for test message:", 'Spam' if prediction else 'Ham')
```
